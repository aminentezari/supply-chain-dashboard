```python
import pandas as pd

df = pd.read_csv("../data/raw/DataCoSupplyChainDataset.csv", encoding="latin-1")
clean = df.copy()  
```

```python
drop_cols = [
    "Customer Email", "Customer Fname", "Customer Lname",
    "Customer Password", "Customer Street", "Customer Zipcode",
    "Product Description", "Product Image", "Order Zipcode",
]
```

```python
clean = clean.drop(columns=drop_cols)
print(clean.shape)
```

```python
#we convert the order date and shipping date columns to datetime format for easier analysis
clean["order_date"] = pd.to_datetime(clean["order date (DateOrders)"], format="%m/%d/%Y %H:%M")
clean["shipping_date"] = pd.to_datetime(clean["shipping date (DateOrders)"], format="%m/%d/%Y %H:%M")

clean["delay_days"] = clean["Days for shipping (real)"] - clean["Days for shipment (scheduled)"]
clean["is_late"] = (clean["Delivery Status"] == "Late delivery").astype(int)
clean["is_canceled"] = clean["Delivery Status"].eq("Shipping canceled").astype(int)
clean["order_year"] = clean["order_date"].dt.year
clean["order_month"] = clean["order_date"].dt.to_period("M").astype(str)
```

```python
clean.columns = (clean.columns.str.strip().str.lower()
                 .str.replace(r"[^a-z0-9]+", "_", regex=True).str.strip("_"))
print(clean.columns.tolist())
```

```python
print(clean.shape)
print(clean["is_late"].mean().round(3), clean["is_canceled"].sum())
print(clean["delay_days"].value_counts().sort_index())
print(clean.isna().sum().sum())

clean.to_csv("../data/clean/orders_clean.csv", index=False)
```
