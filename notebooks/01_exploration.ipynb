## Step 1 Summary: Key Findings

**Dataset:** DataCo Smart Supply Chain (Kaggle)
- 180,519 rows and 53 columns, 0 duplicate rows
- Only 65,752 unique orders (each row is one order item, so one order can have several rows)
- Order dates: 1 Jan 2015 to 31 Jan 2018
- 5 markets and 23 order regions

**Delivery performance**
- 54.8% of rows are late deliveries (98,977 rows)
- Late % by shipping mode:
  - First Class: 95.3%
  - Second Class: 76.6%
  - Same Day: 45.7%
  - Standard Class: 38.1%

**Main insight**
The faster shipping modes are the most late. The reason is the promised delivery time: First Class promises 1 day but takes about 2, and Second Class promises 2 days but takes about 4. Standard Class promises 4 days and delivers in about 4. The problem is the delivery promise, not the carrier speed.

**Data quality notes**
- `Late_delivery_risk` is 1 only when Delivery Status is "Late delivery", so it is consistent.
- 7,754 rows are "Shipping canceled" (Order Status CANCELED or SUSPECTED_FRAUD). They have no real delivery and will be excluded from delivery analysis.
- 18.7% of rows have negative profit.
- Product Description is 100% empty and Order Zipcode is about 86% empty.
- Personal data columns (email, name, password, street, zipcode) will be removed in Step 2.
- Dates were stored as text and must be converted to real dates.
